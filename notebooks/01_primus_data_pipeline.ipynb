# Phase 1: PrIMuS Data Pipeline & Token Sequence Alignment

This notebook demonstrates the foundational dataset pipeline for custom OMR training using the **PrIMuS** (Printed Images of Music Scores) dataset.

### Pipeline Objectives:
1. Load single-line sheet music score images (`.png`).
2. Parse **Agnostic** (`.agnostic`) token sequences (physical symbol locations relative to stafflines).
3. Parse **Semantic** (`.semantic`) token sequences (musical primitives).
4. Verify visual alignment between score slice images and target token sequences.


```python
# Setup Environment & Imports
import os
import sys
import matplotlib.pyplot as plt

# Add root directory to sys.path
root_dir = os.path.dirname(os.getcwd()) if os.path.basename(os.getcwd()) == "notebooks" else os.getcwd()
if root_dir not in sys.path:
    sys.path.insert(0, root_dir)

from data.primus_loader import PrIMuSDataset

print("[SUCCESS] Setup complete! Dataset imports initialized.")
```

```python
# Instantiate PrIMuS Agnostic Dataset Loader
dataset_agnostic = PrIMuSDataset(data_dir="data/primus", annotation_type="agnostic", sample_count=10)
img, token_ids, raw_tokens = dataset_agnostic[0]

print("=== PRIMUS AGNOSTIC SAMPLE ===")
print("Image Dimensions:", img.size)
print("Raw Agnostic Tokens:", raw_tokens)
print("Vocabulary Size:", len(dataset_agnostic.vocab))
```

```python
# Visualize Score Image & Agnostic Token Alignment
plt.figure(figsize=(14, 3))
plt.imshow(img, cmap="gray")
plt.title(f"PrIMuS Agnostic Tokens ({len(raw_tokens)} symbols):\n" + " | ".join(raw_tokens), fontsize=10, fontweight="bold")
plt.axis("off")
plt.tight_layout()
plt.show()
```

```python
# Instantiate PrIMuS Semantic Dataset Loader
dataset_semantic = PrIMuSDataset(data_dir="data/primus", annotation_type="semantic", sample_count=10)
img_sem, token_ids_sem, raw_tokens_sem = dataset_semantic[0]

print("=== PRIMUS SEMANTIC SAMPLE ===")
print("Image Dimensions:", img_sem.size)
print("Raw Semantic Tokens:", raw_tokens_sem)
print("Vocabulary Size:", len(dataset_semantic.vocab))
```

```python
# Visualize Score Image & Semantic Token Alignment
plt.figure(figsize=(14, 3))
plt.imshow(img_sem, cmap="gray")
plt.title(f"PrIMuS Semantic Tokens ({len(raw_tokens_sem)} primitives):\n" + " | ".join(raw_tokens_sem), fontsize=10, fontweight="bold")
plt.axis("off")
plt.tight_layout()
plt.show()
```
