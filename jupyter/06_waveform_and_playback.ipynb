# Step 6: Waveform Visualization & Audio Playback

This notebook plots the synthesized WAV waveform and provides inline audio playback in Jupyter.


```python
# Setup environment & imports
import os
import sys
import wave
import numpy as np
import matplotlib.pyplot as plt
from IPython.display import Audio, display

backend_path = r"c:\Users\hamza\Desktop\S2S\backend"
if backend_path not in sys.path:
    sys.path.insert(0, backend_path)
project_root = r"c:\Users\hamza\Desktop\S2S"
if project_root not in sys.path:
    sys.path.insert(0, project_root)

wav_input = "empty_core_output.wav"
if not os.path.exists(wav_input):
    print(f"⚠️ Warning: '{wav_input}' not found. Please run Notebook 05 first.")
else:
    print(f"✅ Input WAV audio found: '{wav_input}'")
```

```python
# Plot audio waveform and render playback widget
target_wav = "empty_core_output.wav"

if not os.path.exists(target_wav):
    print(f"Error: Audio file '{target_wav}' does not exist. Run Notebook 05 first.")
else:
    with wave.open(target_wav, "r") as wf:
        frames = wf.readframes(wf.getnframes())
        rate = wf.getframerate()
        audio_data = np.frombuffer(frames, dtype=np.int16)

    time_axis = np.linspace(0, len(audio_data) / rate, num=len(audio_data))

    plt.figure(figsize=(12, 4))
    plt.plot(time_axis, audio_data, color="#5C5470", alpha=0.85)
    plt.title(f"Synthesized Waveform ({target_wav})", fontsize=14, fontweight="bold", color="#352F44")
    plt.xlabel("Time (seconds)", fontsize=12)
    plt.ylabel("Amplitude", fontsize=12)
    plt.grid(True, linestyle="--", alpha=0.5)
    plt.tight_layout()
    plt.show()

    display(Audio(target_wav))
```
