# Step 4: Quantization & MIDI Export

This notebook resolves note ties, quantizes measures, and exports the clean score to standard MIDI (`.mid`).


```python
# Setup environment & imports
import os
import sys
import music21

backend_path = r"c:\Users\hamza\Desktop\S2S\backend"
if backend_path not in sys.path:
    sys.path.insert(0, backend_path)
project_root = r"c:\Users\hamza\Desktop\S2S"
if project_root not in sys.path:
    sys.path.insert(0, project_root)

xml_input = "empty_core_omr.musicxml"
if not os.path.exists(xml_input):
    print(f"⚠️ Warning: '{xml_input}' not found. Please run Notebook 02 first.")
else:
    print(f"✅ Input MusicXML found: '{xml_input}'")
```

```python
# Clean score ties/quantization and export to MIDI
raw_xml_path = "empty_core_omr.musicxml"
midi_output_path = "empty_core_output.mid"

if not os.path.exists(raw_xml_path):
    raise FileNotFoundError(f"Target MusicXML '{raw_xml_path}' not found. Run Notebook 02 first.")

score = music21.converter.parse(raw_xml_path)

try:
    score.makeTies(inPlace=True)
    score.quantize(inPlace=True)
except Exception as e:
    print("Notice during quantization:", e)

score.write("midi", fp=midi_output_path)
print("Cleaned MIDI track exported:", midi_output_path, "size:", os.path.getsize(midi_output_path), "bytes")
```
